```python
status_rate = (
    df.groupby(["payment_method", "status"])
      .size()
      .unstack(fill_value=0)
)

status_rate = (
    status_rate.div(status_rate.sum(axis=1), axis=0) * 100
).round(2)

status_rate = status_rate[["Success", "Refunded", "Charged-back"]]

status_rate.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Transaction Status by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Percentage of Transactions")
plt.xticks(rotation=0)
plt.legend(title="Status")
plt.tight_layout()
plt.show()
```

```python
risk_rate = status_rate[["Refunded", "Charged-back"]]

sns.barplot(
    data=risk_rate.reset_index().melt(
        id_vars="payment_method",
        var_name="Risk Type",
        value_name="Rate"
    ),
    x="payment_method",
    y="Rate",
    hue="Risk Type"
)

plt.title("Refund & Chargeback Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
status_value = (df.groupby(['payment_method', 'status'])['amount'].sum().unstack(fill_value = 0))
status_value
```

```python
risk_value = status_value[["Refunded", "Charged-back"]].reset_index()

risk_value = risk_value.melt(
    id_vars="payment_method",
    var_name="risk_type",
    value_name="amount"
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=risk_value,
    x="payment_method",
    y="amount",
    hue="risk_type"
)

plt.title("Refund & Chargeback Value by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Transaction Value (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
loss_impact = (
    df.groupby("payment_method")
      .agg(
          processed_volume=("amount", "sum"),
          refund_loss=("refund_loss", "sum"),
          chargeback_loss=("chargeback_loss", "sum"),
          net_revenue=("net_revenue", "sum")
      )
)

loss_impact["refund_loss_pct"] = (
    loss_impact["refund_loss"] / loss_impact["processed_volume"] * 100
).round(3)

loss_impact["chargeback_loss_pct"] = (
    loss_impact["chargeback_loss"] / loss_impact["processed_volume"] * 100
).round(3)

loss_impact["total_risk_loss_pct"] = (
    (loss_impact["refund_loss"] + loss_impact["chargeback_loss"])
    / loss_impact["processed_volume"] * 100
).round(3)

loss_impact.sort_values("total_risk_loss_pct", ascending=False)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=loss_impact.reset_index(),
    x="payment_method",
    y="total_risk_loss_pct"
)

plt.title("Risk Loss as % of Processed Volume")
plt.xlabel("Payment Method")
plt.ylabel("Refund + Chargeback Loss (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
profitability = (
    df.groupby("payment_method")
      .agg(
          processed_volume=("amount", "sum"),
          gateway_revenue=("gateway_revenue", "sum"),
          net_revenue=("net_revenue", "sum")
      )
)

profitability["net_take_rate"] = (
    profitability["net_revenue"]
    / profitability["processed_volume"]
    * 100
).round(3)

profitability = profitability.sort_values(
    "net_take_rate",
    ascending=False
)

profitability
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=profitability.reset_index(),
    x="payment_method",
    y="net_take_rate"
)

plt.title("Net Take Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Net Revenue per ₹100 Processed")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit = (
    df.groupby("merchant_category")
      .agg(
          processed_volume=("amount", "sum"),
          net_revenue=("net_revenue", "sum"),
          transactions=("amount", "count")
      )
)

category_profit["net_take_rate"] = (
    category_profit["net_revenue"]
    / category_profit["processed_volume"]
    * 100
).round(3)

category_profit = category_profit.sort_values(
    "net_revenue",
    ascending=False
)

category_profit
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_profit.reset_index(),
    x="merchant_category",
    y="net_revenue"
)

plt.title("Net Revenue by Merchant Category")
plt.xlabel("Merchant Category")
plt.ylabel("Net Revenue (₹)")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_profit.reset_index(),
    x="merchant_category",
    y="net_take_rate"
)

plt.title("Net Take Rate by Merchant Category")
plt.xlabel("Merchant Category")
plt.ylabel("Net Revenue per ₹100 Processed")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
monthly = (
    df.groupby("month")
      .agg(
          transactions=("transaction_id", "count"),
          processed_volume=("amount", "sum"),
          net_revenue=("net_revenue", "sum")
      )
)

monthly
```

```python
merchant_performance = (
    df.groupby(["merchant_id", "merchant_name"])
      .agg(
          transactions=("transaction_id", "count"),
          processed_volume=("amount", "sum"),
          net_revenue=("net_revenue", "sum")
      )
      .sort_values("processed_volume", ascending=False)
)

merchant_performance.head(10)
```

```python
merchant_performance["volume_share"] = (
    merchant_performance["processed_volume"]
    / merchant_performance["processed_volume"].sum()
    * 100
).round(2)

merchant_performance.head(10)
```

```python
merchant_performance["net_take_rate"] = (
    merchant_performance["net_revenue"]
    / merchant_performance["processed_volume"]
    * 100
).round(3)

merchant_performance.sort_values(
    "net_take_rate",
    ascending=False
).head(10)
```
